```python
# UIII Act 4 Diccionarios utilizando funciones

# Sistema de Supervisor de Línea de Producción


# Diccionario vacío
diccionario_supervisor = {}

# --------------------------------------------------
# Función agregar supervisor
def agregar_supervisor():

    clave = input("Ingresa la clave del supervisor: ")

    nombre = input("Ingresa el nombre del supervisor: ")

    linea = input("Ingresa la línea de producción: ")

    empleados = input("Ingresa empleados bajo mando: ")

    incidencias = input("Ingresa reporte de incidencias: ")

    turno = input("Ingresa turno de trabajo: ")

    calidad = input("Ingresa indicador de calidad: ")

    planta = input("Ingresa planta: ")

    diccionario_supervisor[clave] = {
        "Nombre": nombre,
        "Linea": linea,
        "Empleados": empleados,
        "Incidencias": incidencias,
        "Turno": turno,
        "Calidad": calidad,
        "Planta": planta
    }

    print("Supervisor agregado correctamente.")

# --------------------------------------------------
# Función listar supervisores
def listar_supervisores():

    print("\n=========== TABLA DE SUPERVISORES ===========")

    for clave, valor in diccionario_supervisor.items():

        print("------------------------------------------------------------")
        print("Clave:", clave)

        for key, value in valor.items():

            print(key, ":", value)

# --------------------------------------------------
# Función buscar supervisor
def buscar_supervisor():

    clave = input("Ingresa la clave del supervisor a buscar: ")

    if clave in diccionario_supervisor:

        print("\nSupervisor encontrado")

        print("--------------------------------------------------")

        for key, value in diccionario_supervisor[clave].items():

            print(key, ":", value)

    else:

        print("Supervisor no encontrado.")

# --------------------------------------------------
# Función borrar supervisor
def borrar_supervisor():

    clave = input("Ingresa la clave del supervisor a borrar: ")

    if clave in diccionario_supervisor:

        del diccionario_supervisor[clave]

        print("Supervisor eliminado correctamente.")

    else:

        print("Supervisor no encontrado.")

# --------------------------------------------------
# Función actualizar supervisor
def actualizar_supervisor():

    clave = input("Ingresa la clave del supervisor a actualizar: ")

    if clave in diccionario_supervisor:

        nombre = input("Nuevo nombre: ")

        linea = input("Nueva línea de producción: ")

        empleados = input("Nuevos empleados bajo mando: ")

        incidencias = input("Nuevo reporte de incidencias: ")

        turno = input("Nuevo turno de trabajo: ")

        calidad = input("Nuevo indicador de calidad: ")

        planta = input("Nueva planta: ")

        diccionario_supervisor[clave] = {
            "Nombre": nombre,
            "Linea": linea,
            "Empleados": empleados,
            "Incidencias": incidencias,
            "Turno": turno,
            "Calidad": calidad,
            "Planta": planta
        }

        print("Supervisor actualizado correctamente.")

    else:

        print("Supervisor no encontrado.")

# --------------------------------------------------
# Menú principal
while True:

    print("\n=========== MENU SUPERVISOR ===========")
    print("1. Agregar supervisor en diccionario")
    print("2. Listar diccionarios de supervisor")
    print("3. Buscar supervisor en diccionario")
    print("4. Borrar supervisor en diccionario")
    print("5. Actualizar supervisor en diccionario")
    print("6. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":

        agregar_supervisor()

        listar_supervisores()

    elif opcion == "2":

        listar_supervisores()

    elif opcion == "3":

        buscar_supervisor()

        listar_supervisores()

    elif opcion == "4":

        borrar_supervisor()

        listar_supervisores()

    elif opcion == "5":

        actualizar_supervisor()

        listar_supervisores()

    elif opcion == "6":

        print("Saliendo del sistema...")
        break

    else:

        print("Opción inválida.")
```
