```python
# 13/mayo/2026

# Supervisor:
# Línea de producción, empleados bajo mando,
# reporte de incidencias, turno de trabajo,
# indicadores de calidad y planta.

# ======================================================
# 🏭 CBTis 128
# 📚 Programación Orientada a Objetos 2026
# 🤖 Especialidad: Inteligencia Artificial
# 👨‍🏫 Docente: Eliseo Nava
# 👨‍🎓 Alumno: Axel Guzman NL = 34
# 🏭 Proyecto: Diccionario de Supervisores
# ======================================================

# =========================
# 🏭 Diccionario vacío
# =========================
diccionario_supervisor = {}

# =========================
# ➕ Agregar supervisor
# =========================
def agregar_supervisor():

    clave = input("Ingresa clave del supervisor: ")
    linea = input("Ingresa línea de producción: ")
    empleados = input("Ingresa empleados bajo mando: ")
    incidencias = input("Ingresa reporte de incidencias: ")
    turno = input("Ingresa turno de trabajo: ")
    calidad = input("Ingresa indicador de calidad: ")
    planta = input("Ingresa planta: ")

    diccionario_supervisor[clave] = {

        "Linea": linea,
        "Empleados": empleados,
        "Incidencias": incidencias,
        "Turno": turno,
        "Calidad": calidad,
        "Planta": planta
    }

    print("\n✅ Supervisor agregado correctamente")

    listar_supervisores()

# =========================
# 📜 Listar supervisores
# =========================
def listar_supervisores():

    print("\n==============================================================================================================================")
    print(f"{'CLAVE':<15}{'LINEA':<20}{'EMPLEADOS':<15}{'INCIDENCIAS':<30}{'TURNO':<15}{'CALIDAD':<12}{'PLANTA':<20}")
    print("==============================================================================================================================")

    for key, value in diccionario_supervisor.items():

        print(f"{key:<15}"
              f"{value['Linea']:<20}"
              f"{value['Empleados']:<15}"
              f"{value['Incidencias']:<30}"
              f"{value['Turno']:<15}"
              f"{value['Calidad']:<12}"
              f"{value['Planta']:<20}")

    print("==============================================================================================================================")

# =========================
# 🔍 Buscar supervisor
# =========================
def buscar_supervisor():

    buscar = input("Ingresa clave del supervisor: ")

    if buscar in diccionario_supervisor:

        print("\n==============================================================================================================================")
        print(f"{'CLAVE':<15}{'LINEA':<20}{'EMPLEADOS':<15}{'INCIDENCIAS':<30}{'TURNO':<15}{'CALIDAD':<12}{'PLANTA':<20}")
        print("==============================================================================================================================")

        value = diccionario_supervisor[buscar]

        print(f"{buscar:<15}"
              f"{value['Linea']:<20}"
              f"{value['Empleados']:<15}"
              f"{value['Incidencias']:<30}"
              f"{value['Turno']:<15}"
              f"{value['Calidad']:<12}"
              f"{value['Planta']:<20}")

        print("==============================================================================================================================")

    else:
        print("❌ Supervisor no encontrado")

# =========================
# ❌ Borrar supervisor
# =========================
def borrar_supervisor():

    borrar = input("Ingresa clave del supervisor a borrar: ")

    if borrar in diccionario_supervisor:

        del diccionario_supervisor[borrar]

        print("\n✅ Supervisor eliminado correctamente")

        listar_supervisores()

    else:
        print("❌ Supervisor no encontrado")

# =========================
# ✏️ Actualizar supervisor
# =========================
def actualizar_supervisor():

    actualizar = input("Ingresa clave del supervisor: ")

    if actualizar in diccionario_supervisor:

        linea = input("Nueva línea de producción: ")
        empleados = input("Nuevos empleados bajo mando: ")
        incidencias = input("Nuevo reporte de incidencias: ")
        turno = input("Nuevo turno de trabajo: ")
        calidad = input("Nuevo indicador de calidad: ")
        planta = input("Nueva planta: ")

        diccionario_supervisor[actualizar] = {

            "Linea": linea,
            "Empleados": empleados,
            "Incidencias": incidencias,
            "Turno": turno,
            "Calidad": calidad,
            "Planta": planta
        }

        print("\n✅ Supervisor actualizado correctamente")

        listar_supervisores()

    else:
        print("❌ Supervisor no encontrado")

# =========================
# 📌 Menú principal
# =========================
while True:

    print("\n=========== MENÚ SUPERVISOR 🏭 ===========")
    print("1. Agregar supervisor")
    print("2. Listar supervisores")
    print("3. Buscar supervisor")
    print("4. Borrar supervisor")
    print("5. Actualizar supervisor")
    print("6. Salir")

    try:
        opcion = int(input("Selecciona una opción: "))

        if opcion == 1:
            agregar_supervisor()

        elif opcion == 2:
            listar_supervisores()

        elif opcion == 3:
            buscar_supervisor()

        elif opcion == 4:
            borrar_supervisor()

        elif opcion == 5:
            actualizar_supervisor()

        elif opcion == 6:
            print("🏭 Programa finalizado")
            break

        else:
            print("❌ Opción no válida")
    except ValueError:
        print("❌ Entrada inválida. Por favor, introduce un número.")

```
